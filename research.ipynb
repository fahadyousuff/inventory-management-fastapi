This notebook will be used for researching and making new tables in the sqlite database that we will use

```python
import sqlite3
```

```python
conn = sqlite3.connect('inventorysystem.db')
```

creating the user Master data

```python
conn.execute("""
create table users(
Id INTEGER PRIMARY KEY AUTOINCREMENT,
usercode varchar(25),
name varchar(40)
)
""")

conn.execute("""
create table product(
Id INTEGER PRIMARY KEY AUTOINCREMENT,
productcode varchar(25),
name varchar(40),
stock int
)
""")

conn.commit()
conn.close()
```

```python
import pandas as pd
```

```python
users = pd.read_sql("select * from users", conn)
product = pd.read_sql("select * from product", conn)
```
